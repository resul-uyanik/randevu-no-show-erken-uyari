```python
import pandas as pd
import numpy as np

# Veriyi oku
df = pd.read_csv('KaggleV2-May-2016.csv')

# Satır ve sütun sayısını kontrol et
print("Veri Boyutu (Satır, Sütun):", df.shape)

# İlk 5 satırı gör
df.head()
```

```python
# 1. Sütun isimlerini daha standart ve hatasız hale getirelim
df.rename(columns={
    'Hipertension': 'Hypertension',
    'Handcap': 'Handicap',
    'No-show': 'NoShow'
}, inplace=True)

# 2. Tarih formatlarını düzeltelim
df['ScheduledDay'] = pd.to_datetime(df['ScheduledDay'])
df['AppointmentDay'] = pd.to_datetime(df['AppointmentDay'])

# ScheduledDay saat bilgisi içerir, AppointmentDay ise sadece gün içerir.
# Gün farkını doğru almak için saatleri sıfırlayarak sadece tarih kısımlarını alalım:
df['ScheduledDate'] = df['ScheduledDay'].dt.normalize()
df['AppointmentDate'] = df['AppointmentDay'].dt.normalize()

# 3. Özellik Mühendisliği: Bekleme Süresi (Lead Time)
# Randevu tarihi ile randevunun alındığı tarih arasındaki gün farkı:
df['LeadTime'] = (df['AppointmentDate'] - df['ScheduledDate']).dt.days

# 4. Hatalı Kayıtları Filtreleme
# - Yaşı negatif olanlar (örn. -1 girilenler)
# - Randevu günü, randevunun alındığı günden önce olan mantıksız kayıtlar (LeadTime < 0)
df = df[(df['Age'] >= 0) & (df['LeadTime'] >= 0)]

# 5. Ek Zaman Özellikleri Türetme
df['AppointmentDayOfWeek'] = df['AppointmentDate'].dt.day_name() # Haftanın günü (Monday, Tuesday vb.)
df['IsSameDay'] = (df['LeadTime'] == 0).astype(int)              # Randevu aynı gün mü alındı?

# 6. Hedef Değişkeni Binary (0/1) Yapma
# 'Yes' -> 1 (Gelmeyenler / No-show), 'No' -> 0 (Gelenler)
df['NoShow'] = df['NoShow'].map({'Yes': 1, 'No': 0})

# Kontrol çıktıları
print("Temizlenmiş Veri Boyutu:", df.shape)
print("\nNo-Show Dağılımı:")
print(df['NoShow'].value_counts(normalize=True).round(3) * 100)
print("\nÖrnek Veri:")
df[['ScheduledDate', 'AppointmentDate', 'LeadTime', 'AppointmentDayOfWeek', 'IsSameDay', 'NoShow']].head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Görselleştirme stili
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# 1. SMS Bildirimi ve Gelmeme Oranı
sms_noshow = df.groupby('SMS_received')['NoShow'].mean().reset_index()
sns.barplot(x='SMS_received', y='NoShow', data=sms_noshow, ax=axes[0], palette='Blues_d')
axes[0].set_title('SMS Hatırlatması vs No-Show Oranı')
axes[0].set_ylabel('Gelmeme Oranı')
axes[0].set_xticklabels(['SMS Yok (0)', 'SMS Var (1)'])

# 2. Haftanın Günleri ve Gelmeme Oranı
order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday']
day_noshow = df.groupby('AppointmentDayOfWeek')['NoShow'].mean().reindex(order).reset_index()
sns.barplot(x='AppointmentDayOfWeek', y='NoShow', data=day_noshow, ax=axes[1], palette='Purples_d')
axes[1].set_title('Haftanın Günü vs No-Show Oranı')
axes[1].set_ylabel('Gelmeme Oranı')
axes[1].tick_params(axis='x', rotation=45)

# 3. Bekleme Süresi (Lead Time) Dağılımı (Kısa vs Uzun Bekleme)
# 0 gün, 1-3 gün, 4-7 gün, 8-15 gün, 16-30 gün, 30+ gün gruplaması
bins = [-1, 0, 3, 7, 15, 30, 200]
labels = ['Aynı Gün (0)', '1-3 Gün', '4-7 Gün', '8-15 Gün', '16-30 Gün', '30+ Gün']
df['LeadTimeGroup'] = pd.cut(df['LeadTime'], bins=bins, labels=labels)

lead_noshow = df.groupby('LeadTimeGroup', observed=True)['NoShow'].mean().reset_index()
sns.barplot(x='LeadTimeGroup', y='NoShow', data=lead_noshow, ax=axes[2], palette='Reds_d')
axes[2].set_title('Bekleme Süresi vs No-Show Oranı')
axes[2].set_ylabel('Gelmeme Oranı')
axes[2].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, roc_auc_score, confusion_matrix
import joblib

# 1. Kullanılacak Sütunları Seçme
features = [
    'Gender', 'Age', 'Scholarship', 'Hypertension', 
    'Diabetes', 'Alcoholism', 'Handicap', 'SMS_received', 
    'LeadTime', 'AppointmentDayOfWeek', 'IsSameDay'
]

X = df[features].copy()
y = df['NoShow']

# 2. Kategorik Değişkenleri Dönüştürme (One-Hot Encoding)
X = pd.get_dummies(X, columns=['Gender', 'AppointmentDayOfWeek'], drop_first=True)

# 3. Train-Test Ayrımı (%80 Eğitim, %20 Test)
# stratify=y parametresi ile train ve testteki no-show oranlarını eşit tutuyoruz
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

# 4. Modeli Tanımlama ve Eğitme
# class_weight='balanced': Gelmeyen azınlık sınıfın ağırlığını artırır
rf_model = RandomForestClassifier(
    n_estimators=100, 
    max_depth=10, 
    class_weight='balanced', 
    random_state=42, 
    n_jobs=-1
)
rf_model.fit(X_train, y_train)

# 5. Tahminler
y_pred = rf_model.predict(X_test)
y_prob = rf_model.predict_proba(X_test)[:, 1] # Risk olasılıkları (0 ile 1 arası)

# 6. Sonuçları Raporlama
print("--- MODEL DEĞERLENDİRME RAPORU ---")
print(classification_report(y_test, y_pred, target_names=['Gelen (0)', 'Gelmeyen (1)']))
print(f"ROC-AUC Skoru: {roc_auc_score(y_test, y_prob):.4f}")

# 7. Modeli ve Sütun Yapısını Kaydetme (Streamlit arayüzünde kullanmak için)
joblib.dump(rf_model, 'no_show_rf_model.pkl')
joblib.dump(list(X.columns), 'model_features.pkl')
print("\nModel ve öznitelik listesi başarıyla kaydedildi!")
```

```python
# Özellik önem düzeylerini görselleştirme
importances = pd.Series(rf_model.feature_importances_, index=X.columns)
plt.figure(figsize=(10, 6))
importances.sort_values().plot(kind='barh', color='teal')
plt.title('Tahminde En Etkili Faktörler (Feature Importance)')
plt.xlabel('Önem Skoru')
plt.show()
```
